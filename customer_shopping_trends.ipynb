# Consumer Shopping Behavior Analysis

## 🎯 Business Problem Statement

A leading retail company wants to better understand its customers' shopping behavior in order to improve sales, customer satisfaction, and long-term loyalty.  
The management team has noticed changes in purchasing patterns across demographics, product categories, and sales channels (online vs. offline). 

They want understanding of which factors, such as **discounts**, **reviews**, **seasons**, or **payment preferences** drive consumer decisions and repeat purchases.

---

## ❓ Overarching Business Question

> **"How can the company leverage consumer shopping data to identify trends, improve customer engagement, and optimize marketing and product strategies?"**

---

## 📋 Key Analysis Objectives
1. **Demographic & Pattern Tracking:** Identify buying behaviors across different customer segments.
2. **Channel Performance:** Compare online versus offline purchasing trends.
3. **Feature Impact Analysis:** Quantify how reviews, discounts, and payment methods drive conversion and retention.
4. **Actionable Insights:** Provide data-driven recommendations for marketing and product placement.


## 📝 Dataset & Methodology Notes

- **Data source:** `customer_shopping_behavior_v2.csv`. This project uses a randomized, synthetically generated dataset built to model realistic, non-degenerate relationships. It is **not real transactional data**, and findings should be read as a demonstration of methodology rather than a real business recommendation.
- **Channel data limitation:** the dataset has no explicit online/offline channel field. `shipping_type` is used later as an imperfect proxy (`Store Pickup` vs. everything else)

## Data Analysis (SQL)

### Reusable customer segmentation view

The New / Returning / Loyal thresholds are used across multiple queries below. Defining as a view  for consistency.

```python
%%sql
CREATE OR REPLACE VIEW customer_segments AS
SELECT *,
       CASE
           WHEN previous_purchases = 1 THEN 'New'
           WHEN previous_purchases BETWEEN 2 AND 10 THEN 'Returning'
           ELSE 'Loyal'
       END AS customer_segment
FROM customer;
```

## What is the Customer Lifecycle Distribution?

```python
%%sql
SELECT customer_segment,
       COUNT(*) AS number_of_customers,
       ROUND(100.0 * COUNT(*) / SUM(COUNT(*)) OVER(), 1) AS percent_of_total
FROM customer_segments
GROUP BY customer_segment
ORDER BY number_of_customers DESC;
```

## What is the age group distribution in the Customer Lifecycle?

```python
%%sql
SELECT customer_segment,
       COUNT(CASE WHEN age_group LIKE 'Young Adult%' THEN 1 END) AS young_adult,
       COUNT(CASE WHEN age_group LIKE 'Adult%' THEN 1 END) AS adult,
       COUNT(CASE WHEN age_group LIKE 'Middle-aged%' THEN 1 END) AS middle_aged,
       COUNT(CASE WHEN age_group LIKE 'Senior%' THEN 1 END) AS senior
FROM customer_segments
GROUP BY customer_segment;
```

### What is the revenue contribution for each age group?

```python
%%sql
SELECT age_group,
       COUNT(*) AS customers,
       ROUND(AVG(purchase_amount),2) AS avg_spend,
       SUM(purchase_amount) AS total_revenue
FROM customer
GROUP BY age_group
ORDER BY total_revenue DESC;
```

### What is the revenue contribution by gender?

```python
%%sql
SELECT gender,
       COUNT(*) AS customers,
       ROUND(AVG(purchase_amount),2) AS avg_spend,
       SUM(purchase_amount) AS total_revenue
FROM customer
GROUP BY gender
ORDER BY total_revenue DESC;
```

### Do subscribers or non-subscribers spend more?

```python
%%sql
SELECT subscription_status,
       COUNT(customer_id) AS total_customers,
       ROUND(100.0 * COUNT(customer_id) / SUM(COUNT(customer_id)) OVER(), 1) AS percent_of_customers,
       ROUND(AVG(purchase_amount),2) AS avg_spend,
       ROUND(SUM(purchase_amount),2) AS total_revenue,
       ROUND(100.0 * SUM(purchase_amount) / SUM(SUM(purchase_amount)) OVER(), 1) AS percent_of_revenue
FROM customer
GROUP BY subscription_status
ORDER BY total_revenue desc
```

### Are repeat customers (more than 5 purchases) likely to subscribe?

```python
%%sql
SELECT subscription_status,
       COUNT(customer_id) AS repeat_customers,
       ROUND(100.0 * COUNT(customer_id) / SUM(COUNT(customer_id)) OVER(), 1) AS percent_of_customers
FROM customer
WHERE previous_purchases > 5
GROUP BY subscription_status
```

### Do any groups spend more when using a discount than the average non-discount amount?

```python
%%sql
WITH global_baseline AS (
    SELECT age_group,
           ROUND(AVG(purchase_amount), 2) AS baseline_avg_spend
    FROM customer
    WHERE discount_applied = 'No'
    GROUP BY age_group
)
SELECT c.age_group,
       COUNT(c.customer_id) AS total_customers,
       ROUND(AVG(CASE WHEN c.discount_applied = 'Yes' THEN c.purchase_amount END), 2) AS avg_spend_with_discount,
       b.baseline_avg_spend AS baseline_avg_spend_no_discount,
       ROUND(
         AVG(CASE WHEN c.discount_applied = 'Yes' THEN c.purchase_amount END) - b.baseline_avg_spend, 2
       ) AS difference_vs_baseline
FROM customer c
JOIN global_baseline b ON c.age_group = b.age_group
GROUP BY c.age_group, b.baseline_avg_spend
ORDER BY difference_vs_baseline DESC;
```

### Is that difference statistically significant?


```python
result = %sql SELECT age_group, discount_applied, purchase_amount FROM customer;
sig_df = result.DataFrame()

from scipy import stats

print("Discount vs. non-discount spend significance check, by age group")
print("-" * 78)
for ag, sub in sig_df.groupby('age_group'):
    disc = sub.loc[sub['discount_applied'] == 'Yes', 'purchase_amount']
    base = sub.loc[sub['discount_applied'] == 'No', 'purchase_amount']
    t, p = stats.ttest_ind(disc, base, equal_var=False)
    sig = "significant (p<0.05)" if p < 0.05 else "not significant"
    print(f"{ag:22s} discount avg={disc.mean():6.2f}  baseline avg={base.mean():6.2f}  p={p:.3f}  -> {sig}")
```

## What are the top 2 most purchased items within each age group?

```python
%%sql
WITH item_counts AS (
    SELECT age_group,
           item_purchased,
           COUNT(*) AS total_orders,
           ROW_NUMBER() OVER (PARTITION BY age_group ORDER BY COUNT(*) DESC) AS item_rank
    FROM customer
    GROUP BY age_group, item_purchased
)
SELECT item_rank,
       age_group,
       item_purchased,
       total_orders
FROM item_counts
WHERE item_rank <= 2
ORDER BY age_group ASC, item_rank ASC;
```

## What are the top 5 most purchased products within each category?

```python
%%sql
WITH item_counts AS (
    SELECT category,
           item_purchased,
           COUNT(customer_id) AS total_orders,
           ROW_NUMBER() OVER (PARTITION BY category ORDER BY COUNT(customer_id) DESC) AS item_rank
    FROM customer
    GROUP BY category, item_purchased
)
SELECT item_rank, category, item_purchased, total_orders
FROM item_counts
WHERE item_rank <= 5;
```

## What items have the top by Product Rating?

```python
%%sql
SELECT item_purchased,
       ROUND(AVG(review_rating::numeric),2) as "Average Product Rating"
FROM customer
GROUP BY item_purchased
ORDER BY AVG(review_rating) DESC
limit 5
```

## Which 5 products have the highest percentage of purchases with discounts applied?

```python
%%sql
SELECT item_purchased,
       ROUND(100.0 * SUM(CASE WHEN discount_applied = 'Yes' THEN 1 ELSE 0 END)/COUNT(*),2) AS discount_rate
FROM customer
GROUP BY item_purchased
ORDER BY discount_rate DESC
LIMIT 5;
```

## What Shipping Types have the highest average purchase amount?

```python
%%sql
SELECT shipping_type,
ROUND(AVG(purchase_amount),2) AS "Average Purchase Amount"
FROM customer
GROUP BY shipping_type;
```

### How does revenue vary by season?

```python
%%sql
SELECT season,
       COUNT(*) AS orders,
       ROUND(AVG(purchase_amount),2) AS avg_order_value,
       SUM(purchase_amount) AS total_revenue
FROM customer
GROUP BY season
ORDER BY total_revenue DESC;
```

### How does revenue vary by payment method?

```python
%%sql
SELECT payment_method,
       COUNT(*) AS customers,
       ROUND(AVG(purchase_amount),2) AS avg_spend,
       SUM(purchase_amount) AS total_revenue
FROM customer
GROUP BY payment_method
ORDER BY total_revenue DESC;
```

### Do higher review ratings correlate with customer retention?


```python
%%sql
SELECT customer_segment,
       ROUND(AVG(review_rating::numeric),2) AS avg_review_rating
FROM customer_segments
GROUP BY customer_segment
ORDER BY avg_review_rating DESC;
```

### Do discounts or subscriptions correlate with customer retention?

```python
%%sql
SELECT subscription_status,
       discount_applied,
       ROUND(AVG(previous_purchases),1) AS avg_previous_purchases
FROM customer
GROUP BY subscription_status, discount_applied
ORDER BY avg_previous_purchases DESC;
```

### What are the top 10 locations that generate the most revenue?

```python
%%sql
SELECT location,
       COUNT(*) AS customers,
       SUM(purchase_amount) AS total_revenue
FROM customer
GROUP BY location
ORDER BY total_revenue DESC
LIMIT 10;
```

### Channel performance: in-store pickup vs. shipped

```python
%%sql
SELECT CASE WHEN shipping_type = 'Store Pickup' THEN 'In-Store Pickup' ELSE 'Shipped' END AS channel_proxy,
       COUNT(*) AS orders,
       ROUND(AVG(purchase_amount),2) AS avg_spend
FROM customer
GROUP BY channel_proxy;
```

## 📊 Key Findings & Recommendations

**1. Demographic & pattern tracking**
- The customer base skews heavily toward **Loyal** (11+ previous purchases, ~62% of customers) and **Returning** (~37%), with almost no customers in the **New** (1 previous purchase) bucket (<1%). That imbalance is itself a finding: the current customer base looks retention-heavy with very little visible new-customer acquisition.
- Revenue is essentially flat across age groups (~$60 avg spend in every bracket) and only mildly higher for male customers ($60.78 vs. $59.15 avg) — age and gender alone are weak predictors of spend in this data; segmentation strategy likely needs to lean on behavioral fields (subscription, discount usage) instead of demographics alone.

**2. Channel performance**
- Using the `shipping_type`-based proxy, in-store pickup and shipped orders show similar average spend (~$59.76 vs. ~$60.11).

**3. Feature impact: discounts, reviews, payment methods**
- **Discounts:** discount purchases run slightly higher than the non-discount baseline in every age group (+$1.15 to +$1.89), but this difference is **not statistically significant**.
- **Subscriptions:** subscribers spend meaningfully more on average ($65.05 vs. $58.28) and use discounts more often (57% vs. 35% of orders).
- **Reviews:** average rating is nearly flat across lifecycle stages (Loyal 3.70, Returning 3.69, New 3.60), review rating doesn't show a strong link to retention.
- **Payment methods:** revenue is broadly even across payment methods (all within ~$37K–$40K), with no method standing out as a meaningful driver.

**4. Actionable recommendations**
- Prioritize **subscription growth** 
- Investigate the **near-absence of new customers**
